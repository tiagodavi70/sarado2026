```python
import pandas as pd
```

```python
df = pd.read_csv("datasets/md_bruto.csv", encoding="latin", delimiter=";")
```

```python
df.loc[90:]
```

```python
df["2000"]
```

```python
# apagar as linhas 1, 91 até o final
```

```python
df = df.drop([0,1]) # apagar uma linha só
```

```python
df = df.loc[2:90]
df = df.drop(columns=["Empenhado"]) # apagar uma série de colunas
```

```python
df = df[df["GRUPO DE DESPESA"] != "TOTAL"] # condição lógica = diferente do total fica no df
df = df.reset_index().drop(columns=["index"])
df
```

```python
colunas = list(map(lambda x: str(x), range(2000, 2020))) + ["Dotação atualizada 2020*"]
colunas
```

```python
df[colunas] = df[colunas].replace(r"\.", "", regex=True).replace(r"\,", ".", regex=True).astype(float)
```

```python
df.to_csv("datasets/md_limpo.csv", index=False)
```
