```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("ascombe.csv")
```

```python
df.groupby("dataset").mean()
df.groupby("dataset").var() # variancia
df.groupby("dataset").std() # desvio padrão
```

```python
df_temp = df[df["dataset"] == "I"]
df_temp
```

```python
alt.Chart(df_temp).mark_circle(size=100, color="crimson").encode(
    x="x",
    y="y"
)
```

```python
alt.Chart(df).mark_circle(size=100).encode(
    x="x",
    y="y",
    color="dataset"
)
```

```python
alt.Chart(df).mark_circle(size=100).encode(
    alt.X("x").scale(zero=False),
    alt.Y("y").scale(zero=False)
).facet(
    facet="dataset"
)
```
